## Regular Expressions

We start by loading the required modules and reading program 'policies.c' into variable `source`:

````python
from pathlib import Path
import re

source = Path('policies.c').read_text()
print(source)
````

````python
source = Path('policies.c').read_text()
matches = re.finditer(r"[^a-zA-Z0-9_]?\s*exit\s*\(", source)
matches_as_list = list(matches)
print(f"Matches:\n{matches_as_list}")

if matches_as_list:
    print("Property is violated!")
````
